Program A

```python
%%shell
cat << 'EOF' > seq_file_alloc.c
#include <stdio.h>

int main() {
    int start, length, i;
    printf("Enter Starting Block: ");
    if (scanf("%d", &start) != 1) return 0;
    printf("Enter File Length (Number of Blocks): ");
    scanf("%d", &length);

    printf("\nAllocated Blocks:\n");
    for(i = 0; i < length; i++) {
        printf("%d ", start + i);
    }
    printf("\n");
    return 0;
}
EOF

gcc seq_file_alloc.c -o seq_file_alloc

# Format: <starting_block> <length>
echo "4 5" | ./seq_file_alloc
```

Program B

```python
%%shell
cat << 'EOF' > indexed_file_alloc.c
#include <stdio.h>

int main() {
    int n, indexBlock, blocks[20], i;
    printf("Enter Index Block: ");
    if (scanf("%d", &indexBlock) != 1) return 0;
    printf("Enter Number of Blocks: ");
    scanf("%d", &n);

    printf("Enter Block Numbers:\n");
    for(i = 0; i < n; i++) {
        scanf("%d", &blocks[i]);
    }

    printf("\nIndex Block: %d\n", indexBlock);
    printf("Allocated Blocks: ");
    for(i = 0; i < n; i++) {
        printf("%d ", blocks[i]);
    }
    printf("\n");
    return 0;
}
EOF

gcc indexed_file_alloc.c -o indexed_file_alloc

# Format: <index_block> <number_of_blocks> <block_1> <block_2> ...
echo "5 4 7 8 12 15" | ./indexed_file_alloc
```

Program C

```python
%%shell
cat << 'EOF' > linked_file_alloc.c
#include <stdio.h>

int main() {
    int n, blocks[20], i;
    printf("Enter Number of Blocks: ");
    if (scanf("%d", &n) != 1) return 0;
    printf("Enter Block Numbers:\n");
    for(i = 0; i < n; i++) {
        scanf("%d", &blocks[i]);
    }

    printf("\nLinked Allocation:\n");
    for(i = 0; i < n - 1; i++) {
        printf("%d --> ", blocks[i]);
    }
    printf("%d --> NULL\n", blocks[n - 1]);
    return 0;
}
EOF

gcc linked_file_alloc.c -o linked_file_alloc

# Format: <number_of_blocks> <block_1> <block_2> ...
echo "4 10 23 45 60" | ./linked_file_alloc
```
