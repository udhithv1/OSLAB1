Program A

```python
%%shell
cat << 'EOF' > seq_file_org.c
#include <stdio.h>

struct student {
    int regno;
    char name[20];
};

int main() {
    FILE *fp;
    struct student s;

    fp = fopen("student.dat", "w");
    printf("Enter Register Number: ");
    if (scanf("%d", &s.regno) != 1) return 0;
    printf("Enter Name: ");
    scanf("%s", s.name);
    fprintf(fp, "%d %s\n", s.regno, s.name);
    fclose(fp);

    fp = fopen("student.dat", "r");
    fscanf(fp, "%d %s", &s.regno, s.name);
    printf("\nRecord Details\n");
    printf("Register Number: %d\n", s.regno);
    printf("Name: %s\n", s.name);
    fclose(fp);

    return 0;
}
EOF

gcc seq_file_org.c -o seq_file_org

# Format: <regno> <name>
echo "101 Alex" | ./seq_file_org
```

```python
%%shell
cat << 'EOF' > direct_file_org.c
#include <stdio.h>

struct student {
    int regno;
    char name[20];
};

int main() {
    FILE *fp;
    struct student s;

    fp = fopen("random.dat", "wb+");
    printf("Enter Register Number: ");
    if (scanf("%d", &s.regno) != 1) return 0;
    printf("Enter Name: ");
    scanf("%s", s.name);

    fwrite(&s, sizeof(s), 1, fp);
    rewind(fp);
    fread(&s, sizeof(s), 1, fp);

    printf("\nRecord Found\n");
    printf("Reg No: %d\n", s.regno);
    printf("Name: %s\n", s.name);
    fclose(fp);

    return 0;
}
EOF

gcc direct_file_org.c -o direct_file_org

# Format: <regno> <name>
echo "102 Jordan" | ./direct_file_org
```

```python
%%shell
cat << 'EOF' > indexed_file_org.c
#include <stdio.h>

struct student {
    int regno;
    char name[20];
};

int main() {
    struct student s[3];
    int key, i;

    printf("Enter 3 Student Records\n");
    for(i = 0; i < 3; i++) {
        scanf("%d %s", &s[i].regno, s[i].name);
    }

    printf("Enter Register Number to Search: ");
    scanf("%d", &key);

    for(i = 0; i < 3; i++) {
        if(s[i].regno == key) {
            printf("\nRecord Found\n");
            printf("Reg No: %d\n", s[i].regno);
            printf("Name: %s\n", s[i].name);
            return 0;
        }
    }

    printf("Record Not Found\n");
    return 0;
}
EOF

gcc indexed_file_org.c -o indexed_file_org

# Format: <reg1> <name1> <reg2> <name2> <reg3> <name3> <search_regno>
echo "101 John 102 Alice 103 Bob 102" | ./indexed_file_org
```
