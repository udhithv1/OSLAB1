Program A

```python
%%shell
cat <<'EOF' > process_creation.c
#include <stdio.h>
#include <stdlib.h>
#include <unistd.h>
#include <sys/wait.h>

int main()
{
    pid_t pid;
    pid = fork();
    if(pid < 0)
    {
        printf("Fork Failed\n");
        exit(1);
    }
    else if(pid == 0)
    {
        printf("\nCHILD PROCESS");
        printf("\nChild PID: %d", getpid());
        printf("\nParent PID: %d\n", getppid());
        exit(0);
    }
    else
    {
        wait(NULL);
        printf("\nPARENT PROCESS");
        printf("\nParent PID: %d", getpid());
        printf("\nParent's Parent PID: %d\n", getppid());
    }
    return 0;
}
EOF
gcc process_creation.c -o process_creation
./process_creation
```

Program B

```python
%%shell
cat <<'EOF' > wait_demo.c
#include <stdio.h>
#include <unistd.h>
#include <sys/wait.h>

int main()
{
    pid_t pid;
    pid = fork();
    if(pid == 0)
    {
        printf("Child Process Running\n");
        sleep(5);
        printf("Child Process Completed\n");
    }
    else
    {
        wait(NULL);
        printf("Parent Resumes Execution\n");
    }
    return 0;
}
EOF
gcc wait_demo.c -o wait_demo
./wait_demo
```
