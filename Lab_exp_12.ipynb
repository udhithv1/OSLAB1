```python
%%shell
cat << 'EOF' > fifo_page.c
#include <stdio.h>

int main() {
    int pages[50], frames[10];
    int n, f, i, j, k = 0;
    int fault = 0, found;

    printf("Enter Number of Pages: ");
    if (scanf("%d", &n) != 1) return 0;

    printf("Enter Reference String:\n");
    for(i = 0; i < n; i++) 
        scanf("%d", &pages[i]);

    printf("Enter Number of Frames: ");
    scanf("%d", &f);

    for(i = 0; i < f; i++)
        frames[i] = -1;

    for(i = 0; i < n; i++) {
        found = 0;
        for(j = 0; j < f; j++) {
            if(frames[j] == pages[i]) {
                found = 1;
                break;
            }
        }
        if(found == 0) {
            frames[k] = pages[i];
            k = (k + 1) % f;
            fault++;
        }
    }

    printf("Total Page Faults = %d\n", fault);
    return 0;
}
EOF

gcc fifo_page.c -o fifo_page

# Provide input values directly via echo:
# Format: <number_of_pages> <reference_string_elements...> <number_of_frames>
echo "7 1 2 3 4 2 1 5 3" | ./fifo_page
```

```python
%%shell
cat << 'EOF' > optimal_page.c
#include <stdio.h>

int main() {
    int pages[50], frames[10];
    int n, f;
    int i, j, k, pos;
    int fault = 0;
    int found;

    printf("Enter Number of Pages: ");
    if (scanf("%d", &n) != 1) return 0;

    printf("Enter Reference String:\n");
    for(i = 0; i < n; i++) 
        scanf("%d", &pages[i]);

    printf("Enter Number of Frames: ");
    scanf("%d", &f);

    for(i = 0; i < f; i++) 
        frames[i] = -1;

    for(i = 0; i < n; i++) {
        found = 0;
        for(j = 0; j < f; j++) {
            if(frames[j] == pages[i]) {
                found = 1;
                break;
            }
        }
        if(found == 0) {
            pos = 0;
            for(j = 0; j < f; j++) {
                if(frames[j] == -1) {
                    pos = j;
                    break;
                }
                int future = 999;
                for(k = i + 1; k < n; k++) {
                    if(frames[j] == pages[k]) {
                        future = k;
                        break;
                    }
                }
                if(j == 0 || future > pos) {
                    pos = j;
                }
            }
            frames[pos] = pages[i];
            fault++;
        }
    }

    printf("Total Page Faults = %d\n", fault);
    return 0;
}
EOF

gcc optimal_page.c -o optimal_page

# Format: <number_of_pages> <reference_string...> <number_of_frames>
echo "7 1 2 3 4 2 1 5 3" | ./optimal_page
```
